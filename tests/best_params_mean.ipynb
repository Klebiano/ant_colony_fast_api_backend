```python
from dotenv import load_dotenv
import os
import urllib
from pathlib import Path
import pandas as pd
from sqlalchemy import create_engine
from IPython.display import display, HTML
```

```python
n_problems = 5
output_filename=r'output\best_params_mean.csv'
best_params_mean_df = pd.read_csv(output_filename)
```

```python
for problem_number in range(1, n_problems + 1):
    # result_df = best_params_mean_df.loc[best_params_mean_df['problem_number'] == problem_number][columns]
    display(HTML(f"<b style='margin-left:10em; font-size: 20px'>---------Problem {problem_number}---------</b>"))
    display((best_params_mean_df.loc[best_params_mean_df['problem_number'] == problem_number]
                .nsmallest(1, columns=['best_path_len_downtime', 'time_to_run_sec'])))
```

```python
for problem_number in range(1, n_problems + 1):
    # result_df = best_params_mean_df.loc[best_params_mean_df['problem_number'] == problem_number][columns]
    display(HTML(f"<b style='margin-left:10em; font-size: 20px'>---------Problem {problem_number}---------</b>"))
    display((best_params_mean_df.loc[best_params_mean_df['problem_number'] == problem_number]['best_path_len_downtime']
             .round(4)
             .value_counts()
             .sort_index(ascending=True)
             .head(5)))
```

```python
for problem_number in range(1, n_problems + 1):
    # result_df = best_params_mean_df.loc[best_params_mean_df['problem_number'] == problem_number][columns]
    display(HTML(f"<b style='margin-left:10em; font-size: 20px'>---------Problem {problem_number}---------</b>"))
    display((best_params_mean_df.loc[best_params_mean_df['problem_number'] == problem_number]
                .nsmallest(20, columns=['best_path_len_downtime']))
                .agg({'time_to_run_sec': 'mean', 'best_path_len_downtime': 'std'}))
```
