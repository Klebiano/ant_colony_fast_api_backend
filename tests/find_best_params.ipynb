```python
from dotenv import load_dotenv
import os
import urllib
from pathlib import Path
import pandas as pd
from sqlalchemy import create_engine
from IPython.display import display, HTML
```

```python
algorithm_used = "genetic"
# output_filename=r'output\results_grid_search_2.csv'
output_filename=r'C:\Users\klebi\OneDrive\Documentos\TCC\offshore_ant_web_dev\backend\tests\output\memetic_algo_results_grid_search.csv'
```

```python
n_problems = 5
final_grid_search_df = pd.read_csv(output_filename)
# final_grid_search_df['tuple_params'] = list(map(tuple, final_grid_search_df[['n_ants', 'alpha', 'beta', 'rho']].values))
```

```python
min_10_smallest_df = pd.DataFrame()
for problem_number in range(1, n_problems + 1):
    min_10_smallest_df = pd.concat([min_10_smallest_df, 
                                    (final_grid_search_df.loc[final_grid_search_df['problem_number'] == problem_number]
                                    .nsmallest(10, columns=['best_path_len_downtime', 'time_to_run_sec']))])
```

```python
min_10_smallest_df
```

```python
if algorithm_used in ["genetic", "memetic"]:
    columns = ['best_path_len_downtime', 'time_to_run_sec', 'mutation_rate', 'population_size', 'number_of_generations']
else:
    columns = ['best_path_len_downtime', 'time_to_run_sec', 'n_ants', 'alpha', 'beta', 'rho']

for problem_number in range(1, n_problems + 1):
    result_df = min_10_smallest_df.loc[min_10_smallest_df['problem_number'] == problem_number][columns]
    display(HTML(f"<b style='margin-left:10em; font-size: 20px'>---------Problem {problem_number}---------</b>"))
    display(result_df)

    if algorithm_used == "genetic":
        display(result_df.agg({'best_path_len_downtime': 'median', 'time_to_run_sec': 'median', 
                                'mutation_rate': lambda x: x.mode().iloc[0], 'population_size': lambda x: x.mode().iloc[0], 'number_of_generations': lambda x: x.mode().iloc[0]}).to_frame().T)
    else:
        display(result_df.agg({'best_path_len_downtime': 'min', 'time_to_run_sec': 'min', 
                    'n_ants': 'mean', 'alpha': 'mean', 'beta': 'mean', 'rho': 'mean'}).to_frame().T)
```
