```python
n_problems = 5
final_grid_search_df = pd.read_csv(output_filename)
# final_grid_search_df['tuple_params'] = list(map(tuple, final_grid_search_df[['n_ants', 'alpha', 'beta', 'rho']].values))
```

```python
min_10_smallest_df = pd.DataFrame()
for problem_number in range(1, n_problems + 1):
    min_10_smallest_df = pd.concat([min_10_smallest_df, 
                                    (final_grid_search_df.loc[final_grid_search_df['problem_number'] == problem_number]
                                    .nsmallest(10, columns=['best_path_len_downtime', 'time_to_run_sec']))])
```

```python
columns = ['best_path_len_downtime', 'time_to_run_sec', 'n_ants', 'alpha', 'beta', 'rho']
for problem_number in range(1, n_problems + 1):
    result_df = min_10_smallest_df.loc[min_10_smallest_df['problem_number'] == problem_number][columns]
    display(HTML(f"<b style='margin-left:10em; font-size: 20px'>---------Problem {problem_number}---------</b>"))
    display(result_df)
    display(result_df.agg({'best_path_len_downtime': 'min', 'time_to_run_sec': 'min', 
                           'n_ants': 'mean', 'alpha': 'mean', 'beta': 'mean', 'rho': 'mean'}).to_frame().T)
```
